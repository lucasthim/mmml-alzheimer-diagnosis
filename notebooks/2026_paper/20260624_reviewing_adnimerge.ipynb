```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

```python
df_adnimerge = pd.read_csv('./../../data/tabular/ADNIMERGE.csv') #,dtype={'PTID': str, 'VISCODE': str, 'IMAGEUID': str})
df_adnimerge.rename(columns={'PTID': 'SUBJECT', 'IMAGEUID': 'IMAGE_DATA_ID'}, inplace=True)
df_adnimerge['IMAGE_DATA_ID'] = 'I' + df_adnimerge['IMAGE_DATA_ID'].fillna(999999).astype(int).astype(str)
print(df_adnimerge.shape)
display(df_adnimerge.head(5))
```

```python
df_selected_mris = pd.read_csv('./../../data/reference/REFERENCE_TABLE_FOR_MRI.csv')
print(df_selected_mris.shape)
display(df_selected_mris.head(5))
```

```python
df_downloaded_mris = pd.read_csv('./../../data/reference/DOWNLOAD_RAW_MRI.csv')
print(df_downloaded_mris.shape)
display(df_downloaded_mris.head(5))
```

```python
df_filtered_mris = df_adnimerge.merge(df_downloaded_mris, on=['SUBJECT', 'IMAGE_DATA_ID'], how='inner')
df_filtered_mris.shape
```
