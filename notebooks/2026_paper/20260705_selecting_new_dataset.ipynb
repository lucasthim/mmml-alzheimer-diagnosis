```python
df_adnimerge = pd.read_csv('./../../data/tabular/ADNIMERGE.csv') #,dtype={'PTID': str, 'VISCODE': str, 'IMAGEUID': str})
df_adnimerge.rename(columns={'PTID': 'SUBJECT', 'IMAGEUID': 'IMAGE_DATA_ID'}, inplace=True)
df_adnimerge['IMAGE_DATA_ID'] = 'I' + df_adnimerge['IMAGE_DATA_ID'].fillna(999999).astype(int).astype(str)
print(df_adnimerge.shape)
display(df_adnimerge.head(5))
```

```python
# Selecting only ADNI3 and ADNI4
print(str(df_adnimerge.query("ORIGPROT.isin(['ADNI3', 'ADNI4']) and IMAGE_DATA_ID != 'I999999'" ).IMAGE_DATA_ID.unique()
.tolist()).strip("[|]").replace("'", ""))
```

```python
new_subjects = df_adnimerge.query("ORIGPROT.isin(['ADNI3', 'ADNI4']) and IMAGE_DATA_ID != 'I999999'").SUBJECT.unique()
old_subjects = df_adnimerge.query("not(ORIGPROT.isin(['ADNI3', 'ADNI4'])) and IMAGE_DATA_ID != 'I999999'").SUBJECT.unique()
```

```python
common_subjects = np.intersect1d(new_subjects, old_subjects)

print(common_subjects.shape)
print(common_subjects)
```
